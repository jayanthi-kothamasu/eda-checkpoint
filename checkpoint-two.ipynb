## Visualize

Create any visualizations for your EDA here. Make note in the form of code comments of what your thought process is for your visualizations.

```python
brand_count = df.groupby("brand").size()     

plt.bar(brand_count.index,brand_count.values,color="blue")
plt.xlabel("Brand Name")
plt.ylabel("No.of Phones")
plt.title(" No.of Phones Per Brand")
## This bar graph shows number of phones  per brand
```

```python


source_counts = df["source"].value_counts()   ##Count no of phones per source

plt.figure(figsize=(6,6))
plt.pie(source_counts.values, labels=source_counts.index, autopct="%1.1f%%", startangle=140)
plt.title("Proportion of Phones Purchased by Source")
plt.show()

### how many phones were bought from each source

```

```python

## This line graph for average feature by brand

avg_features = df.groupby('brand')[['battery_life_rating', 'camera_rating', 'performance_rating']].mean()
avg_features.plot(kind='line', marker='o', figsize=(10,6))
plt.title("Average Feature Ratings by Brand")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.show()
```
